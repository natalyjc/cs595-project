```python
temp_df = temp_df.drop(columns=["year_month", "month", "temp_anomaly_baseline" ])
```

```python
raw_df = emissions_df[["year", "country", "region", "co2_emissions_mt", "population_millions", "co2_per_capita_t",	"co2_intensity_kg_per_gdp_usd"]]
# print(raw_df.duplicated(["country", "year"]).sum())
```

```python
df = raw_df.merge(
    energy_df,
    on=["year", "country", "region"],
    how="left"
)
```

```python
df
```

```python
temp_agg = (
    temp_df
    .groupby(["year", "region"], as_index=False)
    .agg(
        temp_anomaly_c=("temp_anomaly_c", "mean")
    )
)
temp_wide = temp_agg.pivot(
    index="year",
    columns="region",
    values="temp_anomaly_c"
)

temp_wide.columns = [
    f"temp_anomaly_{region}"
    for region in temp_wide.columns
]

temp_wide = temp_wide.reset_index()
df = df.merge(
    temp_wide,
    on="year",
    how="left"
)
```
